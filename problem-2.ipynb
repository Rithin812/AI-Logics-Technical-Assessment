```python
n=int(input())
a= list(map(int, input().split()))
k= int(input())

best= 0
pos=1
for i in range(n):
    minimum= a[i]
    maximum= a[i]
    for j in range(i, n):
        minimum= min(minimum, a[j])
        maximum= max(maximum, a[j])

        if maximum-minimum <= k:
            length = i-j+1
            if length > best:
                best= length
                pos= i+1    
print(best, pos)
```
