```python
n1= int(input())
a=list(map(int, input().split()))

n2=int(input())
b= list(map(int, input().split()))
ans= []
c =0
i= 0

while i <n1 or i<n2 or c:
    x = a[i] if i < n1 else 0
    y = b[i] if i < n2 else 0
    total = x + y + c
    ans.append(total % 10)
    c= total // 10
    i =i+1
print(*ans)
```
