```python
n= int(input())
a =[]

for i in range(n):
    x,y =map(int,input().split())
    a.append([x,y])
a.sort()

start = a[0][0]
end = a[0][1]
for i in range(1, n):
    if a[i][0]<=end:
        end =max(end,a[i][1])
    else:
        print(start, end)
        start= a[i][0]
        end= a[i][1]
print(start,end)
```
